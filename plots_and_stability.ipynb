```python
import control as ctrl
import matplotlib.pyplot as plt

# Plot impulse response
def plot_impulse_response(num, denum):
    G = ctrl.TransferFunction(num, denum)

    time, response = ctrl.impulse_response(G);

    plt.plot(time, response)
    plt.xlabel("Time (s)")
    plt.ylabel("Output")
    plt.title("Impulse Response")

    plt.grid()
    plt.show()

# Bode plot
def plot_bode(num, denum):
    G = ctrl.TransferFunction(num, denum)

    ctrl.bode_plot(G)
    plt.show()

# Plot root locus
def root_locus(num, denum):
    G = ctrl.TransferFunction(num, denum)

    ctrl.root_locus_plot(G)
    plt.show()

# Check stability of system
def is_stable(num, denum):
    G = ctrl.TransferFunction(num, denum)
    poles = ctrl.poles(G)
    print_poles(poles)
    for pole in poles:
        if(pole.real > 0):
            print("Unstable")
            return

    print("Stable")

def print_poles(poles):
    print("Poles: ")
    for pole in poles:
        print(pole)


```

```python
num = [1]
denum = [1, 0.4, 1] # G(s) = 1/(s^2+0.4s+1)

is_stable(num, denum)
plot_impulse_response(num, denum)
plot_bode(num, denum)
root_locus(num, denum)

        
```
